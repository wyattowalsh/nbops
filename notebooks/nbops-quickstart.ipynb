# nbops quickstart

Local-first runtime observability for Python notebooks. Google Colab is the flagship adapter.


```python
# %pip install -q nbops
from pathlib import Path
from nbops import ObserverConfig, observe

_previous = globals().get("observer")
if _previous is not None and _previous.is_running:
    _previous.stop()

config = ObserverConfig(
    project="my-colab-run",
    output_dir=Path("/content/nbops/my-colab-run") if Path("/content").exists() else Path("nbops/my-colab-run"),
    interval_s=2.0,
    collect_gpu=True,
    collect_processes=True,
    persist=True,
    privacy="redacted",
    ui_mode="auto",
)
observer = observe(config=config)
observer.status()
```

```python
observer.display()
```

```python
observer.stop()
html_report = observer.export_report(format="html")
markdown_report = observer.export_report(format="markdown")
bundle = observer.export_bundle()
print(f"HTML report: {html_report}")
print(f"Markdown report: {markdown_report}")
print(f"Bundle: {bundle}")
```
